```python
import pandas as pd
import glob
import os

# Load all price data
price_files = glob.glob('prices_round_3_day_*.csv')
prices_df = pd.concat([pd.read_csv(f, sep=';') for f in price_files], ignore_index=True)

# Load all trades data
trade_files = glob.glob('trades_round_3_day_*.csv')
trades_df = pd.concat([pd.read_csv(f, sep=';') for f in trade_files], ignore_index=True)
```

```python
# Filter for Velvetfruit related products
vf_products = [
    'VELVETFRUIT_EXTRACT',
    'VEV_4000', 'VEV_4500', 'VEV_5000', 'VEV_5100', 'VEV_5200', 
    'VEV_5300', 'VEV_5400', 'VEV_5500', 'VEV_6000', 'VEV_6500'
]

vf_prices = prices_df[prices_df['product'].isin(vf_products)]
vf_trades = trades_df[trades_df['product'].isin(vf_products)]
```

```python
display(vf_prices.head())
display(vf_trades.head())
```

```python
import matplotlib.pyplot as plt

# Filter just the base VELVETFRUIT_EXTRACT
extract_prices = vf_prices[vf_prices['product'] == 'VELVETFRUIT_EXTRACT'].copy()
# Sort by day and timestamp to ensure correct timeline
extract_prices.sort_values(by=['day', 'timestamp'], inplace=True)

# Create the plot
plt.figure(figsize=(15, 6))
plt.plot(range(len(extract_prices)), extract_prices['mid_price'], label='VELVETFRUIT_EXTRACT', color='purple')
plt.title('VELVETFRUIT_EXTRACT Mid Price Across 3 Days', fontsize=14)
plt.xlabel('Time (Ticks)', fontsize=12)
plt.ylabel('Mid Price', fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
